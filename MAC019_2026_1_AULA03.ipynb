```python
# instalação do anastruct
!pip install anastruct

# importando os pacotes
import numpy as np
import matplotlib.pyplot as plt

from anastruct import SystemElements
```

```python
# REPRESENTAÇÃO APOIO 2º GÊNERO

# Criação de um novo sistema estrutural: objeto ss pertencente à classe SystemElements
ss = SystemElements()

ss.add_element(location=[[0,0],[5,1]])
ss.add_support_hinged(node_id=1)
ss.add_support_hinged(node_id=2)
ss.show_structure()

# Versão simplificada (fica subentendido que parte da origem)
ss = SystemElements()

ss.add_element([5, 1])
ss.add_support_hinged(node_id=[1, 2])
ss.show_structure()
```

```python
# REPRESENTAÇÃO APOIO 1º GÊNERO

ss = SystemElements()

ss.add_element(location=[5,5])
ss.add_support_roll(node_id=1, direction='x') # restringe translação em y, pode ser direction=2
ss.add_support_roll(node_id=2, direction='y') # restringe translação em x, pode ser direction=1
ss.show_structure()
```

```python
# REPRESENTAÇÃO ENGASTE

ss = SystemElements()

ss.add_element(location=[0,2.5])
ss.add_support_fixed(node_id=1)
ss.show_structure()
```

```python
# REPRESENTAÇÃO CARGA CONCENTRADA

ss = SystemElements()

ss.add_element(location=[0, 1])
ss.point_load(node_id=2, Fx=10)
#ss.point_load(node_id=2, Fx=10, rotation=45)
#ss.point_load(node_id=2, Fy=10, rotation=45)
ss.show_structure()

# 'rotation' gira a força no sentido horário; a rotação é em graus
```

```python
# REPRESENTAÇÃO CARGA DISTRIBUÍDA

ss = SystemElements()

ss.add_element([5, 0])
ss.q_load(q=+1, element_id=1, direction='x')
ss.q_load(q=-1, element_id=1, direction='x')
#ss.q_load(q=-1, element_id=ss.id_last_element, direction='element') # equivalente
ss.show_structure()
```

```python
# REPRESENTAÇÃO MOMENTO CONCENTRADO

ss = SystemElements()

ss.add_element(location = [5, 0])
ss.moment_load(node_id=ss.id_last_node, Ty=20)
ss.show_structure()

# positivo no sentido anti-horário
```
